```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

sboms_dir = "../../data/results/sboms"
vulns_dir = "../../data/results/vulns"
cicd_dir = "../../data/results/cicd"

dependencias = []
vulnerabilidades = []
cicd_issues = []

for filename in os.listdir(sboms_dir):
    if filename.endswith(".json"):
        with open(os.path.join(sboms_dir, filename), "r") as f:
            data = json.load(f)
            repo_name = filename.replace("_sbom.json", "")
            for artifact in data.get("artifacts", []):
                dependencias.append({
                    "Repositorio": repo_name,
                    "Paquete": artifact.get("name"),
                    "Versión": artifact.get("version"),
                    "Tipo": artifact.get("type")
                })

for filename in os.listdir(vulns_dir):
    if filename.endswith(".json"):
        with open(os.path.join(vulns_dir, filename), "r") as f:
            data = json.load(f)
            repo_name = filename.replace("_vuln.json", "")
            for match in data.get("matches", []):
                vuln = match.get("vulnerability", {})
                artifact = match.get("artifact", {})
                vulnerabilidades.append({
                    "Repositorio": repo_name,
                    "CVE": vuln.get("id"),
                    "Severidad": vuln.get("severity"),
                    "Paquete": artifact.get("name")
                })

for filename in os.listdir(cicd_dir):
    if filename.endswith(".json"):
        with open(os.path.join(cicd_dir, filename), "r") as f:
            data = json.load(f)
            repo_name = data.get("repositorio")
            for hallazgo in data.get("hallazgos", []):
                for issue in hallazgo.get("issues", []):
                    cicd_issues.append({
                        "Repositorio": repo_name,
                        "Workflow": hallazgo.get("workflow"),
                        "Problema": issue
                    })

df_deps = pd.DataFrame(dependencias)
df_vulns = pd.DataFrame(vulnerabilidades)
df_cicd = pd.DataFrame(cicd_issues)

plt.figure(figsize=(10, 6))
if not df_deps.empty:
    top_deps = df_deps['Paquete'].value_counts().head(10)
    sns.barplot(x=top_deps.values, y=top_deps.index, palette="viridis")
    plt.title("Top 10 Dependencias Más Frecuentes")
    plt.xlabel("Apariciones")
    plt.ylabel("Paquete")
    plt.show()

plt.figure(figsize=(10, 6))
if not df_vulns.empty:
    orden_severidad = ["Critical", "High", "Medium", "Low", "Negligible", "Unknown"]
    severidades_presentes = [s for s in orden_severidad if s in df_vulns['Severidad'].unique()]
    ax = sns.countplot(data=df_vulns, x="Severidad", order=severidades_presentes, palette="Reds_r")
    plt.title("Vulnerabilidades por Severidad")
    plt.ylabel("Cantidad")
    for p in ax.patches:
        if p.get_height() > 0:
            ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='bottom')
    plt.show()

plt.figure(figsize=(10, 6))
if not df_cicd.empty:
    ax = sns.countplot(data=df_cicd, y="Problema", palette="magma")
    plt.title("Problemas de Configuración en CI/CD")
    plt.xlabel("Frecuencia")
    plt.ylabel("Tipo de Problema")
    for p in ax.patches:
        if p.get_width() > 0:
            ax.annotate(f'{int(p.get_width())}', (p.get_width(), p.get_y() + p.get_height() / 2.), ha='left', va='center')
    plt.show()
```

## Análisis Cualitativo e Interpretación de Resultados

La evaluación de seguridad realizada sobre los repositorios de la organización FlowiseAI permite establecer una correlación directa con los incidentes de Ejecución Remota de Código (RCE) reportados entre 2025 y 2026. El análisis integrado de las tres dimensiones revela fallas estructurales en la cadena de suministro de software del proyecto.

1. **Vulnerabilidades en Dependencias:** El análisis cuantitativo arrojó la presencia de 14 vulnerabilidades críticas y 170 altas. La concentración de estas vulnerabilidades en el núcleo del sistema facilita escenarios de explotación donde componentes desactualizados o comprometidos actúan como vectores de entrada para RCE en instancias públicas.
2. **Prácticas de CI/CD:** La detección de la ausencia del comando `npm ci` en los flujos automatizados de GitHub Actions representa un riesgo crítico. La falta de bloqueo estricto en el árbol de dependencias (`lockfile`) permite la inyección de versiones mutadas o maliciosas durante los procesos de construcción.
3. **Código Fuente:** La configuración del análisis estático (SAST) mediante CodeQL complementa la detección de fallos en la lógica de negocio y sanitización de variables.

Los problemas identificados no responden exclusivamente a prácticas locales de desarrollo aisladas, sino que reflejan una gestión deficiente del ciclo de vida del software y de la superficie de ataque de las dependencias. La acumulación de deuda técnica en la actualización de librerías y la configuración permisiva en los pipelines de integración son factores determinantes que viabilizan el compromiso de las instancias de Flowise documentadas en el ecosistema actual.
